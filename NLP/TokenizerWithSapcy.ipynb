```python
# Collecting email ids of students from students information sheet.
```

```python
with open("students.txt") as f:
    text = f.readlines()
text
```

```python
# Join the list of lines into a single string using space as separator.
# we do this because nlp() method takes a string as input, usually a long line.
text = " ".join(text)
text
```

```python
# this extracts email ids from the text file.
# Process the text with Spacy NLP pipeline.
doc = nlp(text) # Create a Doc object by processing the text with the nlp pipeline.
emails = []
for token in doc: # Iterate through each token in the Doc object.
    if token.like_email: # Check if the token resembles an email address.
        emails.append(token.text) # If it does, append the token's text to the emails list.
emails        
```

```python
# Customizing tokenizer
```

```python
from spacy.symbols import ORTH

nlp = spacy.blank("en")
doc = nlp("gimme double cheese extra large healthy pizza") # Tokenize the sentence using the default tokenizer.
tokens = [token.text for token in doc] 
tokens
```

```python
# Adding a special case to the tokenizer
nlp.tokenizer.add_special_case("gimme", [
    {ORTH: "gim"}, # split "gimme" into two tokens "gim" and "me" , this creates 2 tokens out of gimme word
    {ORTH: "me"}, 
])
# then tokenize it.
doc = nlp("gimme double cheese extra large healthy pizza")
tokens = [token.text for token in doc]
tokens
```

```python
# Sentence Tokenization or Segmentation
```

```python
nlp.add_pipe("sentencizer") # add sentence boundary detector, which is a rule-based sentence segmentation component that we just added to this blank pipeline.
doc = nlp("Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of delhi")
for sentence in doc.sents: # Iterate through the sentences in the Doc object. (here, .sents is a generator, it iterates over the sentences)
    print(sentence)
```

```python
nlp.pipeline # Check the components in the pipeline
```
