```python
# Text Representation - Bag Of Words (BOW)
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("spam.csv")
df.head()
```

```python
df.Category.value_counts() # this shows how many spam and ham (regular emails) emails we have
```

```python
# Create a new column 'spam' where spam = 1 and ham = 0
df['spam'] = df['Category'].apply(lambda x: 1 if x =='spam' else 0)
```

```python
df.shape
```

```python
df.head()
```

```python
# Train test split
```

```python
from sklearn.model_selection import train_test_split
# our X is the Message column and y is the spam column(1 for spam, 0 for ham)
X_train, X_test, y_train, y_test = train_test_split(df.Message, df.spam, test_size=0.2)
```

```python
# Create bag of words representation using CountVectorizer
```

```python
from sklearn.feature_extraction.text import CountVectorizer

v = CountVectorizer()

X_train_cv = v.fit_transform(X_train.values) # when we use fit_transform, it learns the vocabulary and creates the bag of words model
X_train_cv # this is a sparse matrix representation of the bag of words model
```

```python
X_train_cv.toarray()[:2][0] # first message array representation
```

```python
X_train_cv.shape # (4457, 7740) -> 4457 messages and 7740 unique words in the vocabulary
```

```python
v.get_feature_names_out()[1771] # checking which word is at index 1771 in the vocabulary
```

```python
v.vocabulary_ # this shows the mapping of words to their index in the vocabulary in the bag of words model
```

```python
X_train_np = X_train_cv.toarray() # converting sparse matrix to a numpy array
X_train_np[0] # first message array representation
```

```python
# Train the naive bayes model
```

```python
# training the above bag of words model (X_train_cv) with Multinomial Naive Bayes again.
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_train_cv, y_train)
```

```python
# we have bag of words model (X_train_cv) for training data, now we need to create bag of 
# words model for test data (X_test_cv).
X_test_cv = v.transform(X_test) # we only use transform here because we don't want to learn a new vocabulary, we want to use the existing one.
```

```python
# Evaluate Performance
```

```python
from sklearn.metrics import classification_report

y_pred = model.predict(X_test_cv) # passing the test bag of words model (X_test_cv) to the trained model to get predictions

print(classification_report(y_test, y_pred))
```

```python
emails = [
    'Hey mohan, can we get together to watch footbal game tomorrow?',
    'Upto 20% discount on parking, exclusive offer just for you. Dont miss this reward!'
]

emails_count = v.transform(emails) # creating bag of words model for the new emails using the existing vocabulary
model.predict(emails_count) # passing the new emails bag of words model to the trained model (we again trained it using NB algorithm above) to get predictions
```

```python
# Train the model using sklearn pipeline and reduce number of lines of code
```

```python
from sklearn.pipeline import Pipeline
# here we are creating a pipeline with two steps: first step is CountVectorizer to create bag of words model, 
# second step is MultinomialNB to train the model
clf = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('nb', MultinomialNB())
])
# in return we got a clf object with bag of words model again trained on NB algorithm.
```

```python
X_train
```

```python
y_train
```

```python
clf.fit(X_train, y_train) # training the pipeline model with training data
```

```python
y_pred = clf.predict(X_test)

print(classification_report(y_test, y_pred))
```
