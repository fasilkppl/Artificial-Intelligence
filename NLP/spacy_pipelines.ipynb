```python
# Blank nlp pipeline
```

```python
import spacy

nlp = spacy.blank("en")

doc = nlp("Captain america ate 100$ of samosa. Then he said I can do this all day.")

for token in doc:
    print(token)
```

```python
nlp.pipeline ## Empty list indicating no components in the pipeline
```

```python
nlp.pipe_names # Empty array indicating no components in the pipeline
```

```python
# nlp.pipe_names is empty array indicating no components in the pipeline. 
# Pipeline is something that starts with a tokenizer 
```

```python
'''Download trained pipeline
To download trained pipeline use a command such as,
python -m spacy download en_core_web_sm
This downloads the small (sm) pipeline for english language
'''
```

```python
nlp = spacy.load("en_core_web_sm") # Load the trained pipeline, which includes components such as tagger, parser, NER etc.
nlp.pipe_names 
```

```python
nlp.pipeline # List of tuples indicating components in the pipeline
```

```python
# sm in en_core_web_sm means small. There are other models available as well such as medium, large etc.
```

```python
doc = nlp("Captain america ate 100$ of samosa. Then he said I can do this all day.")

for token in doc:
    print(token, " | ", spacy.explain(token.pos_), " | ", token.lemma_) # Print token, part-of-speech and lemma.
# spacy.explain(token.pos_)  this gives the explanation of the part-of-speech tag
# token.lemma_ gives base form of the word. eg ate -> eat.
```

```python
# Named Entity Recognition. 
# NER identifies and classifies named entities in text 
# into predefined categories such as persons, organizations, locations, dates, etc.
# eg Apple was founded by Steve Jobs in California. Here, Apple is ORG, Steve Jobs is PERSON, California is GPE (Geopolitical Entity)
```

```python
'''
NER (Named Entity Recognition) is an NLP task that identifies and classifies important real-world entities in text. 
NER finds “names” in text and tells you what kind they are.
'''
```

```python
doc = nlp("Tesla Inc is going to acquire twitter for $45 billion")
for ent in doc.ents:
    print(ent.text, " | ", ent.label_, " | ", spacy.explain(ent.label_))

# ent.label_ gives the label of the entity such as ORG, MONEY etc.
# You can use spacy.explain(ent.label_) to get the explanation of the label.
```

```python
from spacy import displacy
# Visualize the named entities in the doc above.
displacy.render(doc, style="ent")
```

```python
# Adding a component to a blank pipeline
```

```python
source_nlp = spacy.load("en_core_web_sm")

nlp = spacy.blank("en")
nlp.add_pipe("ner", source=source_nlp)
nlp.pipe_names
```

```python
doc = nlp("Tesla Inc is going to acquire twitter for $45 billion")
for ent in doc.ents:
    print(ent.text, ent.label_)
```
