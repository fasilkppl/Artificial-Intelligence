```python
# Sentence & Word Tokenization In Spacy
```

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")

doc = nlp("Dr. Strange loves pav bhaji of mumbai. Hulk loves chaat of delhi")
```

```python
for sentence in doc.sents: # here doc.sents is a generator that yields sentences one by one.
    print(sentence)
```

```python
# Word Tokenization : this time we will loop through each sentence (splits it) and then each word in that sentence
for sentence in doc.sents:
    for word in sentence:
        print(word)
```
