```python
# Text Classification Using Spacy Word Embeddings - GloVe
```

```python
'''
Problem Statement

- Fake news refers to misinformation or disinformation in the country which is spread through 
word of mouth and more recently through digital communication such as What's app messages, social media posts, etc.

- Fake news spreads faster than real news and creates problems and fear among groups and in society.

- We are going to address these problems using classical NLP techniques and going to classify whether 
a given message/ text is **Real or Fake Message**.

- We will use **glove embeddings** from spacy which is trained on massive wikipedia dataset to 
pre-process and text vectorization and apply different classification algorithms.



- This data consists of two columns.
        - Text
        - label
- Text is the statements or messages regarding a particular event/situation.

- label feature tells whether the given text is Fake or Real.

- As there are only 2 classes, this problem comes under the **Binary Classification.**
'''
```

```python
import pandas as pd

#read the dataset with name "Fake_Real_Data.csv" and store it in a variable df
df = pd.read_csv("Fake_Real_Data.csv")

#print the shape of dataframe
print(df.shape)

#print top 5 rows
df.head(5)
```

```python
#check the distribution of labels 
df['label'].value_counts()
```

```python
# From the above, we can see that almost the labels(classes) occured equal number of times and balanced.
# There is no problem of class imbalance and hence no need to apply any balancing techniques like undersampling, oversampling etc.
```

```python
#Add the new column which gives a unique number to each of these labels 
df['label_num'] = df['label'].map({'Fake' : 0, 'Real': 1})

#check the results with top 5 rows
df.head(5)
```

```python
# Get spacy word vectors and store them in a pandas dataframe
```

```python
import spacy
nlp = spacy.load("en_core_web_lg") # loading spacy large model
```

```python
#This will take some time.
#  creating a column names vector where vectors for each sentence/ word are stored.
df['vector'] = df['Text'].apply(lambda text: nlp(text).vector)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df.vector.values, # get values inside the vector arrays
    df.label_num,
    test_size=0.2,
    random_state=2022
)
```

```python
# Attempt 1:
import numpy as np
# coverting X_train & X_test to 2D array because our MultinomialNB onlu takes 2d array as input.
# MultinomialNB does require a 2D array of shape (n_samples, n_features)
X_train_2d = np.stack(X_train)
X_test_2d = np.stack(X_test)
```

```python
X_train
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import MinMaxScaler

# our X_train_2d & X_test_2d currently have negative values, so we use scaler to change them into positive.
scaler = MinMaxScaler()
scaled_train_embed = scaler.fit_transform(X_train_2d)
scaled_test_embed = scaler.transform(X_test_2d) # As this is testing data, we only need to do transform.

# if we give negative values to this model, it will show error
clf = MultinomialNB()
clf.fit(scaled_train_embed, y_train)
```

```python
from sklearn.metrics import classification_report

y_pred = clf.predict(scaled_test_embed)

print(classification_report(y_test, y_pred))
```

```python
# Attempt 2:
from  sklearn.neighbors import KNeighborsClassifier

#1. creating a KNN model object
clf = KNeighborsClassifier(n_neighbors = 5, metric = 'euclidean')

#2. fit with all_train_embeddings and y_train, here for KNN negative values in vectors are not a problem.
clf.fit(X_train_2d, y_train)

#3. get the predictions for all_test_embeddings and store it in y_pred
y_pred = clf.predict(X_test_2d)

#4. print the classfication report
print(classification_report(y_test, y_pred))
```

```python
'''
✅ For KNN:
Negative values are NOT a problem
KNN works by computing distances (Euclidean, cosine, etc.)
Distance calculations handle negative numbers just fine

-------------------------------------------------------------

❌ For Multinomial Naive Bayes (MultinomialNB):
Negative values ARE a problem
MultinomialNB models counts or frequencies
Internally uses log probabilities
Requires non-negative feature values

'''
```

```python
# Confusion Matrix
```

```python
#finally print the confusion matrix for the best model
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm

from matplotlib import pyplot as plt
import seaborn as sn
plt.figure(figsize = (10,7))
sn.heatmap(cm, annot=True, fmt='d')
plt.xlabel('Prediction')
plt.ylabel('Truth')
```

```python
'''Key Takeaways

1. KNN model which didn't perform well in the vectorization techniques like Bag of words, and 
TF-IDF due to very **high dimensional vector space**, performed really well with glove vectors due to only 
**300-dimensional** vectors and very good embeddings(similar and related words have almost similar embeddings) 
for the given text data.

2. MultinomialNB model performed decently well but did not come into the top list because in the 
300-dimensional vectors we also have the negative values present. The Naive Bayes model does not fit the data if there are **negative values**. 
So, to overcome this shortcoming, we have used the **Min-Max scaler** to bring down all the values between 0 to 1. 
In this process, there will be a possibility of variance and information loss among the data. 
But anyhow we got a decent recall and f1 scores.

'''
```
