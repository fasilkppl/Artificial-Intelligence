```python
nlp = spacy.load("en_core_web_sm")

doc = nlp("We just opened our wings, the flying part is coming soon")

for token in doc:
    if token.is_stop:
        print(token)
```

```python
# Preprocessing function to remove stop words from text
def preprocess(text):
    doc = nlp(text)
    # List comprehension to filter out stop words, here token.is_stop is used to check if the token is a stop word
    no_stop_words = [token.text for token in doc if not token.is_stop]  # if token is not a stop word then enter inside loop and add to list.
    return " ".join(no_stop_words) # Join the list of words back into a single string
```

```python
preprocess("Musk wants time to prepare for a trial over his")
```

```python
def preprocess(text):
    doc = nlp(text)
    # List comprehension to filter out stop words, here token.is_stop is used to check if the token is a stop word
    no_stop_words = [token.text for token in doc if not token.is_stop and not token.is_punct]  # if token is not a stop word and not punctuation then enter inside loop and add to list.
    return no_stop_words
```

```python
preprocess("The other is? not other but! your, divine brother")
```

```python
# Dataset contains press releases of different court cases from depart of justice (DOJ).
# The releases contain information such as outcomes of criminal cases, notable actions taken against felons, or other updates about the current administration.
```

```python
import pandas as pd

df = pd.read_json("doj_press.json",lines=True) # here lines=True indicates that each line of the file is a separate JSON object.

df.shape
```

```python
df.head(5)
```

```python
# above dataset have null values in topics column.
# Filter out those rows that do not have any topics associated with the case
```

```python
df = df[df["topics"].str.len() != 0] # keep only those rows where length of topics column is not equal to zero.
df.head()
```

```python
df.shape # after removing rows with null topics shape has changed.
```

```python
df =df.head(100) # for demonstration purpose taking only first 100 rows.
df.shape
```

```python
df.head(2)
```

```python
df["contents_new"] = df.contents.apply(preprocess) # here preprocess function filters out stop words from contents column and creates a new column contents_new.
```
